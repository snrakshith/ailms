## P01 — model artifact
Dependencies: scikit-learn joblib.


```python
from sklearn.datasets import load_iris
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
import tempfile,joblib
from pathlib import Path
X,y=load_iris(return_X_y=True)
model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=300)).fit(X[10:],y[10:])
with tempfile.TemporaryDirectory() as tmp:
    path=Path(tmp)/'trusted-local-model.joblib'; joblib.dump(model,path)
    restored=joblib.load(path) # only trusted artifacts; pickle can execute code
    assert (restored.predict(X[:10])==model.predict(X[:10])).all()
print('serialization parity passed; this is not a quality benchmark')
```

## P02 — drift monitoring
Dependencies: numpy.


```python
import numpy as np
rng=np.random.default_rng(3)
base=rng.normal(0,1,1000); new=rng.normal(.8,1,1000)
shift=abs(new.mean()-base.mean())/base.std()
print('standardized mean shift',shift,'investigate',shift>.5)
# Threshold is illustrative: calibrate on historical windows and monitor labels separately.
```
