## F01 — python data
Dependencies: numpy pandas.


```python
import numpy as np
import pandas as pd
x = np.array([[1., 2., 5.], [3., 4., 5.]])
scale = x.std(axis=0)
z = (x - x.mean(axis=0)) / np.where(scale == 0, 1, scale)
assert np.isfinite(z).all()
assert np.allclose(z[:, :2].mean(axis=0), 0)
frame = pd.DataFrame({'group':['A','B','A'], 'score':[1,2,None]})
print(z, frame.groupby('group').score.mean(), sep='\n')
```

## F02 — math basics
Dependencies: numpy.


```python
import numpy as np
p = .9*.01 / (.9*.01 + .05*.99)
assert abs(p - .153846) < 1e-5
w = 0.
for _ in range(50): w -= .1 * 2*(w-3)
assert abs(w-3) < .001
A = np.array([[1.,2.],[3.,4.]])
u,s,vt = np.linalg.svd(A, full_matrices=False)
assert np.allclose((u*s)@vt,A)
print('Posterior',p,'optimum',w)
```
