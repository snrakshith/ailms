Reverse a valid 1-based inclusive sublist.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def reverse_between(head,left,right):
    dummy=ListNode(0,head);before=dummy
    for _ in range(left-1):before=before.next
    tail=before.next
    for _ in range(right-left):
        node=tail.next;tail.next=node.next;node.next=before.next;before.next=node
    return dummy.next

assert reverse_between(ListNode(1,ListNode(2,ListNode(3))),1,2)==ListNode(2,ListNode(1,ListNode(3)))
```
