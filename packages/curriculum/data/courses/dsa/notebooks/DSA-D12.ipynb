Find any peak index in nonempty data with distinct neighbors.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def peak(a):
    if not a:raise ValueError('empty')
    l,r=0,len(a)-1
    while l<r:
        m=(l+r)//2
        if a[m]<a[m+1]:l=m+1
        else:r=m
    return l

assert peak([1,4,3,2])==1 and peak([5])==0
```
