# DSA-D48
Count structurally distinct BSTs on n ordered keys.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def catalan(n):
    dp=[1]+[0]*n
    for i in range(1,n+1):dp[i]=sum(dp[j]*dp[i-j-1] for j in range(i))
    return dp[n]

assert catalan(3)==5 and catalan(0)==1
```
