Reconstruct a tree from consistent unique traversals.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def build_tree(pre,ino):
    pos={x:i for i,x in enumerate(ino)};i=0
    def build(l,r):
        nonlocal i
        if l>=r:return None
        x=pre[i];i+=1;m=pos[x]
        return TreeNode(x,build(l,m),build(m+1,r))
    return build(0,len(ino))

assert build_tree([2,1,3],[1,2,3])==TreeNode(2,TreeNode(1),TreeNode(3)) and build_tree([],[]) is None
```
