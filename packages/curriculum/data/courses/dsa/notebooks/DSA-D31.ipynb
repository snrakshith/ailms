Compute distance from nearest source in an adjacency list.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def multi_bfs(g,sources):
    d={s:0 for s in sources};q=deque(d)
    while q:
        u=q.popleft()
        for v in g[u]:
            if v not in d:d[v]=d[u]+1;q.append(v)
    return d

assert multi_bfs([[1],[0,2],[1,3],[2,4],[3]],[0,4])=={0:0,4:0,1:1,3:1,2:2}
```
