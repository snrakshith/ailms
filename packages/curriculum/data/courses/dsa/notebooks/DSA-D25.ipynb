Validate a strict binary search tree.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def valid_bst(root):
    def walk(n,l,r):return n is None or (l<n.val<r and walk(n.left,l,n.val) and walk(n.right,n.val,r))
    return walk(root,-inf,inf)

assert not valid_bst(TreeNode(5,TreeNode(3,None,TreeNode(6)))) and valid_bst(TreeNode(2,TreeNode(1),TreeNode(3)))
```
