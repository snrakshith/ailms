Validate a string of bracket characters.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def brackets(s):
    stack=[];pairs={')':'(',']':'[','}':'{'}
    for c in s:
        if c in pairs.values():stack.append(c)
        elif c not in pairs or not stack or stack.pop()!=pairs[c]:return False
    return not stack

assert brackets('([])') and not brackets('([)]') and brackets('')
```
