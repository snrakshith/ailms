# DSA-D58
Find first missing positive integer, mutating input.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def missing_positive(a):
    n=len(a)
    for i in range(n):
        while 1<=a[i]<=n and a[a[i]-1]!=a[i]:
            j=a[i]-1;a[i],a[j]=a[j],a[i]
    for i,x in enumerate(a,1):
        if x!=i:return i
    return n+1

assert missing_positive([3,4,-1,1])==2 and missing_positive([1,1])==2 and missing_positive([])==1
```
