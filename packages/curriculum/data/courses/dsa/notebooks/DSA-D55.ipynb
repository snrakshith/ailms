# DSA-D55
Maximize count of courses with durations and deadlines.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def schedule_courses(a):
    heap=[];total=0
    for duration,deadline in sorted(a,key=lambda x:x[1]):
        total+=duration;heappush(heap,-duration)
        if total>deadline:total+=heappop(heap)
    return len(heap)

assert schedule_courses([(3,3),(2,4),(1,5)])==2 and schedule_courses([])==0
```
