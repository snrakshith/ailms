# DSA-D34
Compute shortest distances, leaving unreachable nodes infinite.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def dijkstra(g,s):
    if any(w<0 for row in g for _,w in row):raise ValueError('negative weight')
    d=[inf]*len(g);d[s]=0;heap=[(0,s)]
    while heap:
        cost,u=heappop(heap)
        if cost!=d[u]:continue
        for v,w in g[u]:
            if cost+w<d[v]:d[v]=cost+w;heappush(heap,(d[v],v))
    return d

assert dijkstra([[(1,5),(2,1)],[],[(1,1)],[]],0)==[0,2,1,inf]
```
