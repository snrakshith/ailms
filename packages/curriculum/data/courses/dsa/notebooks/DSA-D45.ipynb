# DSA-D45
Count down/right paths in grid with obstacles.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def grid_paths(g):
    if not g or not g[0]:return 0
    dp=[1]+[0]*(len(g[0])-1)
    for row in g:
        for j,x in enumerate(row):
            if x:dp[j]=0
            elif j:dp[j]+=dp[j-1]
    return dp[-1]

assert grid_paths([[0,0],[0,0]])==2 and grid_paths([[1]])==0
```
