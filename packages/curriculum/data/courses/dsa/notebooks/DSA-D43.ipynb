# DSA-D43
Return best total value within integer capacity.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def knapsack(weights,values,W):
    dp=[0]*(W+1)
    for w,v in zip(weights,values):
        for cap in range(W,w-1,-1):dp[cap]=max(dp[cap],dp[cap-w]+v)
    return dp[W]

assert knapsack([2,3],[3,4],4)==4 and knapsack([2],[3],4)==3
```
