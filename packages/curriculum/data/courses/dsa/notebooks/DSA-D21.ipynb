Compute tree depth using recursion.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def depth(root):return 0 if root is None else 1+max(depth(root.left),depth(root.right))

assert depth(TreeNode(1,TreeNode(2)))==2 and depth(None)==0
```
