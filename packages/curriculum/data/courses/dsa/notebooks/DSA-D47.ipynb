Maximize unlimited-transaction profit with fee and no overlapping holdings.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def stock_fee(a,fee):
    held=-inf;cash=0
    for x in a:held,cash=max(held,cash-x),max(cash,held+x-fee)
    return cash

assert stock_fee([1,3,2,8,4,9],2)==8 and stock_fee([],2)==0
```
