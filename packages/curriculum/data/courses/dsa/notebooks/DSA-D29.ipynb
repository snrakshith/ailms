Maintain online median for nonempty integer stream.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

class median_stream:
    def __init__(self):self.lo=[];self.hi=[]
    def add(self,x):
        heappush(self.lo,-x);heappush(self.hi,-heappop(self.lo))
        if len(self.hi)>len(self.lo):heappush(self.lo,-heappop(self.hi))
    def median(self):
        if not self.lo:raise ValueError('empty')
        return -self.lo[0] if len(self.lo)>len(self.hi) else (-self.lo[0]+self.hi[0])/2

m=median_stream();m.add(3);assert m.median()==3;m.add(1);assert m.median()==2;m.add(2);assert m.median()==2
```
