# DSA-D27
Maximum independent-set weight on a binary tree, permitting empty selection.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def rob_tree(root):
    def walk(n):
        if not n:return 0,0
        a,b=walk(n.left);c,d=walk(n.right)
        return n.val+b+d,max(a,b)+max(c,d)
    return max(walk(root))

assert rob_tree(TreeNode(3,TreeNode(4),TreeNode(5)))==9 and rob_tree(TreeNode(-1))==0
```
