# DSA-D53
Decide if last index is reachable with nonnegative jump limits.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def can_jump(a):
    reach=0
    for i,x in enumerate(a):
        if i>reach:return False
        reach=max(reach,i+x)
    return bool(a)

assert not can_jump([3,2,1,0,4]) and can_jump([2,0,1])
```
