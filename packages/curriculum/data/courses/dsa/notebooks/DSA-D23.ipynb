Return whether any root-to-leaf path equals target.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def path_sum(root,t):
    if root is None:return False
    t-=root.val
    if root.left is None and root.right is None:return t==0
    return path_sum(root.left,t) or path_sum(root.right,t)

assert path_sum(TreeNode(1,TreeNode(2)),3) and not path_sum(TreeNode(1,TreeNode(2)),1)
```
