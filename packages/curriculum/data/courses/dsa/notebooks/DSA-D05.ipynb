# DSA-D05
Compute largest histogram rectangle.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def histogram(a):
    stack=[]; best=0
    for i,h in enumerate(a+[0]):
        while stack and a[stack[-1]]>h:
            height=a[stack.pop()]; left=stack[-1] if stack else -1
            best=max(best,height*(i-left-1))
        stack.append(i)
    return best

assert histogram([2,1,2])==3 and histogram([2,2])==4 and histogram([])==0
```
