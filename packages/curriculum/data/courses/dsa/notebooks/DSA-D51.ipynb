# DSA-D51
Find minimum walk length visiting every node in an undirected graph.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def visit_all(g):
    n=len(g)
    if not n:return 0
    seen={(i,1<<i) for i in range(n)};q=deque((i,m,0) for i,m in seen);full=(1<<n)-1
    while q:
        u,m,d=q.popleft()
        if m==full:return d
        for v in g[u]:
            state=(v,m|(1<<v))
            if state not in seen:seen.add(state);q.append((*state,d+1))
    return -1

assert visit_all([[1],[0,2],[1]])==2 and visit_all([[],[]])==-1
```
