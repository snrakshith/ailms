Compute integer power for nonnegative exponent.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def power(a,n):
    if n<0:raise ValueError('nonnegative exponent required')
    out=1
    while n:
        if n&1:out*=a
        a*=a;n>>=1
    return out

assert power(3,13)==1594323 and power(9,0)==1
```
