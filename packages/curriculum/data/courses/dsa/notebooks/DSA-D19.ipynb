# DSA-D19
Merge k sorted lists of integers into a new list.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def merge_k(lists):
    heap=[];out=[]
    for j,a in enumerate(lists):
        if a:heappush(heap,(a[0],j,0))
    while heap:
        x,j,i=heappop(heap);out.append(x)
        if i+1<len(lists[j]):heappush(heap,(lists[j][i+1],j,i+1))
    return out

assert merge_k([[1,4],[2,3],[]])==[1,2,3,4] and merge_k([])==[]
```
