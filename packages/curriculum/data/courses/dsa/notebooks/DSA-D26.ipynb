Count good nodes in a tree.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def good_nodes(root):
    def walk(n,m):
        if not n:return 0
        return int(n.val>=m)+walk(n.left,max(m,n.val))+walk(n.right,max(m,n.val))
    return walk(root,-inf)

assert good_nodes(TreeNode(3,TreeNode(1),TreeNode(4)))==2 and good_nodes(None)==0
```
