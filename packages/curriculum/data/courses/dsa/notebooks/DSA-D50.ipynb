Compute binary-tree diameter in edges.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def diameter(root):
    best=0
    def walk(n):
        nonlocal best
        if not n:return 0
        l,r=walk(n.left),walk(n.right);best=max(best,l+r);return 1+max(l,r)
    walk(root);return best

assert diameter(TreeNode(1,TreeNode(2,TreeNode(3))))==2 and diameter(None)==0
```
