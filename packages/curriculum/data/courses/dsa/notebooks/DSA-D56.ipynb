Find one unpaired integer using XOR.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def single_number(a):
    x=0
    for n in a:x^=n
    return x

assert single_number([4,1,4,2,2])==1 and single_number([-2])==-2
```
