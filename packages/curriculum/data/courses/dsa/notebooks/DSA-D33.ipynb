# DSA-D33
Return a topological order or None if cyclic.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def topo(n,edges):
    g=[[] for _ in range(n)];degree=[0]*n
    for a,b in edges:g[a].append(b);degree[b]+=1
    q=deque(i for i,x in enumerate(degree) if x==0);out=[]
    while q:
        u=q.popleft();out.append(u)
        for v in g[u]:
            degree[v]-=1
            if degree[v]==0:q.append(v)
    return out if len(out)==n else None

assert topo(3,[(0,1),(1,2)])==[0,1,2] and topo(2,[(0,1),(1,0)]) is None
```
