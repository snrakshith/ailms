# DSA-D59
Maximize profit from nonoverlapping positive-duration jobs.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def job_profit(jobs):
    jobs=sorted(jobs,key=lambda x:x[1]);ends=[x[1] for x in jobs];dp=[0]
    for i,(s,e,p) in enumerate(jobs):
        j=bisect_right(ends,s,0,i);dp.append(max(dp[-1],p+dp[j]))
    return dp[-1]

assert job_profit([(1,3,5),(2,5,7),(3,6,8)])==13 and job_profit([])==0
```
