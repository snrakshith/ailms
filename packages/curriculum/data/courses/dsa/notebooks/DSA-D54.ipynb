Merge closed intervals into sorted disjoint intervals.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def merge_intervals(a):
    out=[]
    for l,r in sorted(a):
        if out and l<=out[-1][1]:out[-1][1]=max(out[-1][1],r)
        else:out.append([l,r])
    return out

assert merge_intervals([[1,3],[2,4],[6,7]])==[[1,4],[6,7]] and merge_intervals([])==[]
```
