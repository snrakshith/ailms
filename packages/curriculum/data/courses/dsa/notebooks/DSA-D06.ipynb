# DSA-D06
Return zero-based indices in sorted input or None.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def sorted_pair(a,target):
    l,r=0,len(a)-1
    while l<r:
        s=a[l]+a[r]
        if s==target:return l,r
        if s<target:l+=1
        else:r-=1
    return None

assert sorted_pair([1,3,5,8],8)==(1,2) and sorted_pair([2],4) is None
```
