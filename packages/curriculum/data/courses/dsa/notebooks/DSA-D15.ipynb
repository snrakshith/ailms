Return number of days until a warmer temperature.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def warmer(a):
    stack=[];out=[0]*len(a)
    for i,x in enumerate(a):
        while stack and a[stack[-1]]<x:
            j=stack.pop();out[j]=i-j
        stack.append(i)
    return out

assert warmer([73,74,72,76])==[1,2,1,0] and warmer([5,5])==[0,0]
```
