# DSA-D02
Count subarrays with sum k, including negative elements.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def subarray_count(a,k):
    seen=Counter({0:1}); total=ans=0
    for x in a:
        total+=x; ans+=seen[total-k]; seen[total]+=1
    return ans

assert subarray_count([2,-2,2],2)==3 and subarray_count([0,0],0)==3
```
