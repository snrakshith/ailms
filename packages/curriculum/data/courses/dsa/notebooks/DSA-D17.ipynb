Reverse a singly linked list represented by node objects.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def reverse_list(head):
    prev=None
    while head:
        nxt=head.next;head.next=prev;prev=head;head=nxt
    return prev

assert reverse_list(ListNode(1,ListNode(2))).val==2 and reverse_list(None) is None
```
