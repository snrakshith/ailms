Maximize burst-balloon score for nonnegative values.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def burst(a):
    b=[1]+a+[1];n=len(b);dp=[[0]*n for _ in b]
    for length in range(2,n):
        for l in range(n-length):
            r=l+length;dp[l][r]=max(dp[l][k]+dp[k][r]+b[l]*b[k]*b[r] for k in range(l+1,r))
    return dp[0][-1]

assert burst([3,1,5,8])==167 and burst([])==0
```
