Maximum sum of nonadjacent values, allowing empty selection.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def rob_line(a):
    prev2=prev=0
    for x in a:prev2,prev=prev,max(prev,prev2+x)
    return prev

assert rob_line([2,7,9,3,1])==12 and rob_line([-1])==0
```
