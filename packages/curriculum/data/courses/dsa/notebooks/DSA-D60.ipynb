# DSA-D60
Return k most frequent integers ordered by descending frequency then ascending value.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def top_frequency(a,k):
    c=Counter(a)
    if not 0<=k<=len(c):raise ValueError('invalid k')
    return sorted(c,key=lambda x:(-c[x],x))[:k]

assert top_frequency([4,4,3,3,2],1)==[3] and top_frequency([],0)==[]
```
