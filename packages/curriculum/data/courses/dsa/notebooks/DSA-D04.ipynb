# DSA-D04
Rotate a square matrix clockwise without another matrix.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def rotate(a):
    n=len(a)
    if any(len(row)!=n for row in a): raise ValueError('square matrix required')
    for i in range(n):
        for j in range(i+1,n): a[i][j],a[j][i]=a[j][i],a[i][j]
    for row in a: row.reverse()
    return a

assert rotate([[1,2],[3,4]])==[[3,1],[4,2]] and rotate([])==[]
```
