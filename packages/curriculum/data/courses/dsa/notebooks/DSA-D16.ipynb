Compute sum of minimums of all nonempty subarrays, exact integer.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def sum_mins(a):
    stack=[];total=0
    for i in range(len(a)+1):
        while stack and (i==len(a) or a[stack[-1]]>=a[i]):
            j=stack.pop();l=stack[-1] if stack else -1
            total+=a[j]*(j-l)*(i-j)
        stack.append(i)
    return total

assert sum_mins([2,2])==6 and sum_mins([3,1,2,4])==17 and sum_mins([])==0
```
