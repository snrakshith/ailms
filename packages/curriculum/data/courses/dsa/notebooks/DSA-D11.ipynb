# DSA-D11
Find minimum integer speed with sum ceil(pile/speed) <= hours.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def min_speed(a,hours):
    if not a or any(x<=0 for x in a) or hours<len(a):raise ValueError('infeasible or invalid')
    l,r=1,max(a)
    while l<r:
        m=(l+r)//2
        if sum((x+m-1)//m for x in a)<=hours:r=m
        else:l=m+1
    return l

assert min_speed([3,6,7,11],8)==4 and min_speed([1],1)==1
```
