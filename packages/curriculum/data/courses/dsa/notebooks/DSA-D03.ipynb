# DSA-D03
Partition an array containing only 0,1,2 in place.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def sort_colors(a):
    l=i=0; r=len(a)-1
    while i<=r:
        if a[i]==0: a[l],a[i]=a[i],a[l]; l+=1; i+=1
        elif a[i]==2: a[r],a[i]=a[i],a[r]; r-=1
        elif a[i]==1: i+=1
        else: raise ValueError('expected 0, 1, 2')
    return a

assert sort_colors([2,0,1,2,0])==[0,0,1,2,2] and sort_colors([])==[]
```
