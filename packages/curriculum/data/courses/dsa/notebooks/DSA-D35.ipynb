Find bridges in an undirected simple graph.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def bridges(n,edges):
    g=[[] for _ in range(n)]
    for a,b in edges:g[a].append(b);g[b].append(a)
    tin=[-1]*n;low=[0]*n;clock=0;out=[]
    def dfs(u,p):
        nonlocal clock
        tin[u]=low[u]=clock;clock+=1
        for v in g[u]:
            if v==p:continue
            if tin[v]!=-1:low[u]=min(low[u],tin[v])
            else:
                dfs(v,u);low[u]=min(low[u],low[v])
                if low[v]>tin[u]:out.append(tuple(sorted((u,v))))
    for u in range(n):
        if tin[u]==-1:dfs(u,-1)
    return sorted(out)

assert bridges(4,[(0,1),(1,2),(2,0),(2,3)])==[(2,3)] and bridges(1,[])==[]
```
