Enumerate palindrome partitions.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def pal_parts(s):
    out=[];path=[]
    def dfs(i):
        if i==len(s):out.append(path.copy());return
        for j in range(i+1,len(s)+1):
            p=s[i:j]
            if p==p[::-1]:path.append(p);dfs(j);path.pop()
    dfs(0);return out

assert pal_parts('aab')==[['a','a','b'],['aa','b']] and pal_parts('')==[[]]
```
