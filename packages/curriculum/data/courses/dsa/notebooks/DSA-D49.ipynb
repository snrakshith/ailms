Implement full-string glob matching for literals, ? and *.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def wildcard(s,p):
    dp=[True]+[False]*len(p)
    for j,c in enumerate(p,1):dp[j]=dp[j-1] and c=='*'
    for x in s:
        new=[False]
        for j,c in enumerate(p,1):new.append((new[-1] or dp[j]) if c=='*' else (dp[j-1] and (c=='?' or c==x)))
        dp=new
    return dp[-1]

assert wildcard('abef','a*f') and not wildcard('ab','a?c') and wildcard('','**')
```
