# DSA-D30
Count 4-connected islands without mutating the grid.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def islands(g):
    if not g or not g[0]:return 0
    r,c=len(g),len(g[0]);seen=set();count=0
    for i in range(r):
        for j in range(c):
            if not g[i][j] or (i,j) in seen:continue
            count+=1;q=[(i,j)];seen.add((i,j))
            while q:
                x,y=q.pop()
                for a,b in ((x+1,y),(x-1,y),(x,y+1),(x,y-1)):
                    if 0<=a<r and 0<=b<c and g[a][b] and (a,b) not in seen:seen.add((a,b));q.append((a,b))
    return count

assert islands([[1,0],[0,1]])==2 and islands([])==0
```
