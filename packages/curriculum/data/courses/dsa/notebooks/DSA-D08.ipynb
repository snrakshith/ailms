# DSA-D08
Return maximum sum among windows of size k.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def fixed_window(a,k):
    if not 1<=k<=len(a):raise ValueError('invalid width')
    s=best=sum(a[i] for i in range(k))
    for i in range(k,len(a)):s+=a[i]-a[i-k];best=max(best,s)
    return best

assert fixed_window([1,4,2,8],2)==10 and fixed_window([-4,-2],1)==-2
```
