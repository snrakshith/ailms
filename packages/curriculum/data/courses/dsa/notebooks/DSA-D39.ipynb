# DSA-D39
Test whether a word exists along 4-neighbor cells without reuse.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def word_search(g,word):
    if not word:return True
    if not g or not g[0]:return False
    r,c=len(g),len(g[0]);seen=set()
    def dfs(x,y,i):
        if i==len(word):return True
        if not(0<=x<r and 0<=y<c) or (x,y) in seen or g[x][y]!=word[i]:return False
        seen.add((x,y));ok=any(dfs(a,b,i+1) for a,b in ((x+1,y),(x-1,y),(x,y+1),(x,y-1)));seen.remove((x,y));return ok
    return any(dfs(i,j,0) for i in range(r) for j in range(c))

assert word_search(['AB','CD'],'ABD') and not word_search(['AB','CD'],'ABA')
```
