# DSA-D36
Find shortest 4-neighbor path allowing k blocked-cell entries, with open endpoints.
Run the cells in order. Reference solution is shown after the task.


```python
from collections import Counter, deque, OrderedDict
from bisect import bisect_left, bisect_right
from heapq import heappush, heappop
from dataclasses import dataclass
from math import inf


@dataclass
class ListNode:
    val:int
    next:object=None

@dataclass
class TreeNode:
    val:int
    left:object=None
    right:object=None

def grid_budget(g,k):
    r,c=len(g),len(g[0]);q=deque([(0,0,k,0)]);seen={(0,0,k)}
    while q:
        x,y,b,d=q.popleft()
        if (x,y)==(r-1,c-1):return d
        for a,z in ((x+1,y),(x-1,y),(x,y+1),(x,y-1)):
            if 0<=a<r and 0<=z<c and b>=g[a][z]:
                state=(a,z,b-g[a][z])
                if state not in seen:seen.add(state);q.append((*state,d+1))
    return -1

assert grid_budget([[0,1,0]],1)==2 and grid_budget([[0,1,0]],0)==-1
```
