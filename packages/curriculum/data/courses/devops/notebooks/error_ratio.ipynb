# error_ratio
Compute request error ratio and design a latency histogram query.
Run the cells in order. Reference solution is shown after the task.


```python

def ratio(total,errors):
    if total<=0 or not 0<=errors<=total:return None
    return errors/total
assert ratio(1000,10)==.01 and ratio(0,0) is None and ratio(3,4) is None
```
