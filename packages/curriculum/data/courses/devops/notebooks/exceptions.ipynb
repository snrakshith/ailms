# exceptions
Create a security gate policy with fail-closed scanner errors and time-bound risk exceptions.
Run the cells in order. Reference solution is shown after the task.


```python

from datetime import date
def allows(exception,today):
    return bool(exception.get('owner') and exception.get('ticket') and exception.get('reason') and date.fromisoformat(exception['expires'])>=today)
valid={'owner':'security','ticket':'LAB-1','reason':'temporary sandbox exception','expires':'2026-12-31'}
assert allows(valid,date(2026,12,1)) and not allows(valid,date(2027,1,1)) and not allows({**valid,'owner':''},date(2026,12,1))
```
