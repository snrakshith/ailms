# log_correlation
Parse synthetic JSON logs and locate error requests by correlation ID.
Run the cells in order. Reference solution is shown after the task.


```python

import json
lines=['{"service":"api","level":"ERROR","trace_id":"a"}','{"service":"db","level":"INFO","trace_id":"a"}','{"service":"api","level":"INFO","trace_id":"b"}']
records=[json.loads(x) for x in lines];bad={r['trace_id'] for r in records if r['level']=='ERROR'}
assert len([r for r in records if r['trace_id'] in bad])==2
```
