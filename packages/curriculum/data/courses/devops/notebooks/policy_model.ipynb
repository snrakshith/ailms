# policy_model
Write a deny policy for privileged containers and evaluate safe and unsafe manifests.
Run the cells in order. Reference solution is shown after the task.


```python

# Offline model of a single policy; run actual OPA separately.
def denied(pod):
    containers=pod.get('spec',{}).get('containers',[])
    return [c['name'] for c in containers if c.get('securityContext',{}).get('privileged') is True]
assert denied({'spec':{'containers':[{'name':'demo','securityContext':{'privileged':True}}]}})==['demo']
assert denied({'spec':{'containers':[{'name':'safe'}]}})==[]
```
