# git_revert
Create a disposable Git repository, commit a change, inspect diff and revert it.
Run the cells in order. Reference solution is shown after the task.


```python

import tempfile, subprocess
from pathlib import Path
with tempfile.TemporaryDirectory() as d:
    def git(*args):return subprocess.run(['git',*args],cwd=d,check=True,capture_output=True,text=True).stdout
    git('init','-b','main');git('config','user.name','Learner');git('config','user.email','learner@example.invalid');git('config','commit.gpgsign','false')
    p=Path(d)/'app.txt';p.write_text('v1');git('add','app.txt');git('commit','-m','baseline');p.write_text('v2');git('commit','-am','update');git('revert','--no-edit','HEAD')
    assert p.read_text()=='v1' and len(git('log','--oneline').splitlines())==3
```
