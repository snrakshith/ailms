# release_gate
Apply a release gate to synthetic metrics with both healthy and unhealthy candidates.
Run the cells in order. Reference solution is shown after the task.


```python

def gate(report):
    required=('tests','sast','sca','secrets','image_scan','signature_verified','dast')
    if not all(report.get(k) is True for k in required):return False
    n=report.get('requests',0);errors=report.get('errors',-1)
    return n>=1000 and 0<=errors<=n and errors/n<=.01
ok={k:True for k in ('tests','sast','sca','secrets','image_scan','signature_verified','dast')};ok.update(requests=1000,errors=5)
assert gate(ok) and not gate({**ok,'signature_verified':False}) and not gate({**ok,'requests':2}) and not gate({**ok,'errors':20})
```
