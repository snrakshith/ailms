# artifact_hash
Hash a build artifact and detect a changed copy.
Run the cells in order. Reference solution is shown after the task.


```python

from hashlib import sha256
artifact=b'learning artifact v1';digest=sha256(artifact).hexdigest()
def verify(content,expected):return sha256(content).hexdigest()==expected
assert verify(artifact,digest) and not verify(artifact+b' changed',digest)
```
