# slo_budget
Run a tabletop exercise for an API error spike and calculate budget consumption.
Run the cells in order. Reference solution is shown after the task.


```python

def burn(slo,total,errors):
    if not 0<slo<1 or total<=0 or not 0<=errors<=total:return None
    return (errors/total)/(1-slo)
assert abs(burn(.999,10000,20)-2)<1e-9 and burn(.999,0,0) is None
```
