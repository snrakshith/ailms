# permissions
Create a local temporary config, set owner-only permissions and verify them.
Run the cells in order. Reference solution is shown after the task.


```python

import tempfile, os, stat
from pathlib import Path
with tempfile.TemporaryDirectory() as d:
    p=Path(d)/'config';p.write_text('mode=lab');p.chmod(0o600)
    assert stat.S_IMODE(p.stat().st_mode)==0o600
    p.chmod(0o644);assert stat.S_IMODE(p.stat().st_mode)!=0o600
```
